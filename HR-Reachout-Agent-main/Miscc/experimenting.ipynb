```python
def get_available_bot(tickets, config):
    all_bots = [b["bot_token"] for b in config["Telegram"]["bots"]]
    print(f"[DEBUG] All bots: {all_bots}")
    used_bots = [t.get("bot_token") for t in tickets if "bot_token" in t]
    print(f"[DEBUG] Used bots: {used_bots}")
    for bot in all_bots:
        if bot not in used_bots:
            print(f"[DEBUG] Selected available bot: {bot}")
            return bot
bot_token = get_available_bot(tickets, config)
```
